```python
def basic_calculator():
    print("📘 Basic Calculator")
    num1 = float(input("Enter first number: "))
    operator = input("Choose operation (+, -, *, /): ")
    num2 = float(input("Enter second number: "))

    if operator == '+':
        result = num1 + num2
    elif operator == '-':
        result = num1 - num2
    elif operator == '*':
        result = num1 * num2
    elif operator == '/':
        if num2 != 0:
            result = num1 / num2
        else:
            return "❌ Division by zero not allowed."
    else:
        return "❌ Invalid operation."

    return f"✅ Result: {result}"

# Run
print(basic_calculator())
```

```python
while True:
    print("\n--- Smart Calculator ---")
    print(basic_calculator())
    again = input("Do you want to calculate again? (yes/no): ").lower()
    if again != "yes":
        print("👋 Exiting calculator.")
        break
```

```python
memory = 0

def basic_calculator_with_memory():
    global memory
    print("📘 Calculator with Memory (M+, MR, MC)")
    command = input("Enter command or calculation (M+, MR, MC or normal calc): ").strip()

    if command.upper() == "MR":
        return f"🧠 Memory Recall: {memory}"
    elif command.upper() == "MC":
        memory = 0
        return "🧽 Memory Cleared."
    elif command.upper().startswith("M+"):
        try:
            value = float(command[2:].strip())
            memory += value
            return f"✅ {value} added to memory. Total: {memory}"
        except:
            return "❌ Invalid M+ value."
    else:
        try:
            num1 = float(input("Enter first number: "))
            operator = input("Choose operation (+, -, *, /): ")
            num2 = float(input("Enter second number: "))

            if operator == '+':
                result = num1 + num2
            elif operator == '-':
                result = num1 - num2
            elif operator == '*':
                result = num1 * num2
            elif operator == '/':
                result = num1 / num2
            else:
                return "❌ Invalid operation."

            return f"✅ Result: {result}"
        except:
            return "❌ Something went wrong."

# Run
print(basic_calculator_with_memory())
```

```python
# 🧮 Smart Python Calculator

# Global variables for memory and history
memory = 0
history = []

def basic_calc():
    try:
        num1 = float(input("Enter first number: "))
        operator = input("Choose operation (+, -, *, /): ")
        num2 = float(input("Enter second number: "))

        if operator == '+':
            result = num1 + num2
        elif operator == '-':
            result = num1 - num2
        elif operator == '*':
            result = num1 * num2
        elif operator == '/':
            if num2 != 0:
                result = num1 / num2
            else:
                return "❌ Division by zero not allowed."

        else:
            return "❌ Invalid operator."

        # Save to history
        history.append(f"{num1} {operator} {num2} = {result}")
        return f"✅ Result: {result}"

    except:
        return "❌ Invalid input."

def smart_input_calc():
    query = input("🧠 Type your calculation (e.g. 'add 5 and 2'): ").lower()
    words = query.split()

    try:
        if "add" in query:
            nums = [float(w) for w in words if w.replace('.', '', 1).isdigit()]
            result = nums[0] + nums[1]
        elif "minus" in query or "subtract" in query:
            nums = [float(w) for w in words if w.replace('.', '', 1).isdigit()]
            result = nums[0] - nums[1]
        elif "multiply" in query or "times" in query:
            nums = [float(w) for w in words if w.replace('.', '', 1).isdigit()]
            result = nums[0] * nums[1]
        elif "divide" in query or "by" in query:
            nums = [float(w) for w in words if w.replace('.', '', 1).isdigit()]
            result = nums[0] / nums[1]
        else:
            return "❌ Couldn't understand your input."

        # Save to history
        history.append(f"{query} = {result}")
        return f"✅ Result: {result}"

    except:
        return "❌ Error in smart calculation."

def memory_feature():
    global memory
    cmd = input("Enter memory command (M+, MR, MC): ").upper()

    if cmd == "MR":
        return f"🧠 Memory Recall: {memory}"
    elif cmd == "MC":
        memory = 0
        return "🧽 Memory Cleared."
    elif cmd.startswith("M+"):
        try:
            value = float(cmd[2:].strip())
            memory += value
            return f"✅ Added {value} to memory. Total: {memory}"
        except:
            return "❌ Invalid value for M+."
    else:
        return "❌ Unknown memory command."

def show_history():
    print("📜 Calculation History:")
    if not history:
        print("No calculations yet.")
    else:
        for i, item in enumerate(history, 1):
            print(f"{i}. {item}")

# 📋 Main menu
def smart_calculator():
    while True:
        print("\n🔸 Smart Calculator Menu:")
        print("1. Basic Calculator")
        print("2. Smart Input Calculator")
        print("3. Memory Feature (M+, MR, MC)")
        print("4. Show History")
        print("5. Exit")

        choice = input("Choose option (1-5): ")

        if choice == '1':
            print(basic_calc())
        elif choice == '2':
            print(smart_input_calc())
        elif choice == '3':
            print(memory_feature())
        elif choice == '4':
            show_history()
        elif choice == '5':
            print("👋 Exiting... Thank you!")
            break
        else:
            print("❌ Invalid option. Try again.")

# 🟢 Run the calculator
smart_calculator()
```
